## 3. Implementasikan konsep template matching tanpa menggunakan library OpenCV untuk multiple object

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# --- Load & grayscale ---
img_path = '../images/Object Detection/bahrain.jpg'
tpl_path = '../images/Object Detection/bahrain-template.jpg'

img = np.array(Image.open(img_path).convert('L'), dtype=np.float32)
template = np.array(Image.open(tpl_path).convert('L'), dtype=np.float32)

img_h, img_w = img.shape
tpl_h, tpl_w = template.shape

# --- Normalisasi template ---
tpl_mean = template.mean()
tpl_std = template.std()
if tpl_std == 0:
    raise ValueError("Template has zero standard deviation")
tpl_norm = (template - tpl_mean) / tpl_std

# --- Hitung NCC (brute-force) ---
res_h = img_h - tpl_h + 1
res_w = img_w - tpl_w + 1
result = np.zeros((res_h, res_w), dtype=np.float32)

for y in range(res_h):
    for x in range(res_w):
        region = img[y:y+tpl_h, x:x+tpl_w]
        r_mean = region.mean()
        r_std = region.std()
        if r_std == 0:
            result[y, x] = 0.0
        else:
            region_norm = (region - r_mean) / r_std
            # dot product -> NCC numerator; denom already handled by normalization
            result[y, x] = (region_norm * tpl_norm).sum() / (tpl_h * tpl_w)  # /N to keep in [-1,1]

# normalisasi ke [0,1] (opsional tapi memudahkan threshold)
res_min, res_max = result.min(), result.max()
if res_max - res_min > 0:
    result_norm = (result - res_min) / (res_max - res_min)
else:
    result_norm = result.copy()

# --- Iterative detection + suppression ---
threshold = 0.65   # threshold
detections = []    # list of (x, y, score)

res_copy = result_norm.copy()

while True:
    idx_flat = np.argmax(res_copy)
    y_peak, x_peak = np.unravel_index(idx_flat, res_copy.shape)
    score = res_copy[y_peak, x_peak]
    if score < threshold:
        break
    detections.append((x_peak, y_peak, float(score)))

    y0 = max(0, y_peak - tpl_h // 2)
    y1 = min(res_copy.shape[0], y_peak + tpl_h // 2)
    x0 = max(0, x_peak - tpl_w // 2)
    x1 = min(res_copy.shape[1], x_peak + tpl_w // 2)
    res_copy[y0:y1, x0:x1] = 0.0



fig, ax = plt.subplots(2, 2, figsize=(15, 11))

# Input Image
ax[0][0].imshow(img, cmap='gray')
ax[0][0].set_title('Input image')

# Template Image
ax[0][1].imshow(template, cmap='gray')
ax[0][1].set_title('Template image')

# Matching map
ax[1][0].imshow(result_norm, cmap='hot')
ax[1][0].set_title('Matching map (normalized)')

ax[1][1].imshow(img, cmap='gray')
ax[1][1].set_title('Detected bounding boxes')
for (x, y, s) in detections:
    rect = plt.Rectangle((x, y), tpl_w, tpl_h, edgecolor='b', facecolor='none', linewidth=2)
    ax[1][1].add_patch(rect)
    ax[1][1].text(x, y-6, f'{s:.2f}', color='yellow', fontsize=9, backgroundcolor='black')

plt.tight_layout()
plt.show()

print("Detections (top-left x,y,score):")
for d in detections:
    print(d)
```

## 4. Implementasikan metode Sobel Edge Detection, Canny Edge Detection, dan Laplacian Edge Detection pada OpenCV

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar
img_car_park = cv.imread('../images/Object Detection/car-park.jpg')

# Konversi ke grayscale
gray = cv.cvtColor(img_car_park, cv.COLOR_BGR2GRAY)
```

### a. Sobel Edge Detection

```python
sobelx = cv.Sobel(gray, cv.CV_64F, 1, 0, ksize=3)
sobely = cv.Sobel(gray, cv.CV_64F, 0, 1, ksize=3)
sobel_combined = cv.magnitude(sobelx, sobely)
sobel_combined = cv.convertScaleAbs(sobel_combined)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1) 
plt.imshow(cv.cvtColor(img_car_park, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### b. Canny Edge Detection

```python
canny_edges = cv.Canny(gray, 100, 200)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1) 
plt.imshow(cv.cvtColor(img_car_park, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(canny_edges, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### c. Laplacian Edge Detection

```python
laplacian = cv.Laplacian(gray, cv.CV_64F)
laplacian = cv.convertScaleAbs(laplacian)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1) 
plt.imshow(cv.cvtColor(img_car_park, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 5. Implementasikan metode Harris Corner Detection dan Shi-Tomasi Detection pada OpenCV

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_chess_board = cv.imread('../images/Object Detection/chess-board.jpg')
gray = cv.cvtColor(img_chess_board, cv.COLOR_BGR2GRAY)
```

### a. Harris Corner Detection

```python
gray_float = np.float32(gray)
harris = cv.cornerHarris(gray_float, blockSize=2, ksize=3, k=0.04)
harris = cv.dilate(harris, None)  
img_chess_board_harris = img_chess_board.copy()
img_chess_board_harris[harris > 0.01 * harris.max()] = [0, 255, 0]

plt.figure(figsize=(10, 15))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(img_chess_board, cv.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(img_chess_board_harris, cv.COLOR_BGR2RGB))
plt.title('Harris Corner Detection (hijau)')

plt.tight_layout()
plt.show()
```

### b. Shi-Tomasi Detection

```python
corners = cv.goodFeaturesToTrack(gray, maxCorners=100, qualityLevel=0.01, minDistance=10)
corners = np.int32(corners)
img_chess_board_shi = img_chess_board.copy()
for c in corners:
    x, y = c.ravel()
    cv.circle(img_chess_board_shi, (x, y), 4, (0, 0, 255), -1)


plt.figure(figsize=(10, 15))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(img_chess_board, cv.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(img_chess_board_shi, cv.COLOR_BGR2RGB))
plt.title('Shi-Tomasi Corner Detection (merah)')

plt.tight_layout()
plt.show()
```

## 6. Implementasikan metode Hough Transform pada OpenCV dengan menggunakan gambar sudoku.jpg.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread('../images/Object Detection/sudoku.jpg')
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

edges = cv.Canny(gray, 50, 150, apertureSize=3)

linesP = cv.HoughLinesP(edges, 1, np.pi / 180, threshold=80, minLineLength=100, maxLineGap=10)

img_houghP = img.copy()
if linesP is not None:
    for line in linesP:
        x1, y1, x2, y2 = line[0]
        cv.line(img_houghP, (x1, y1), (x2, y2), (0, 0, 255), 2)  # Garis merah

plt.figure(figsize=(8, 8))
plt.imshow(cv.cvtColor(img_houghP, cv.COLOR_BGR2RGB))
plt.title('Deteksi Garis Sudoku (Hough Transform Probabilistic)')
plt.axis('off')
plt.show()
```

## 7. Implementasikan fungsi findContours() pada OpenCV untuk contour detection dengan menggunakan gambar laptop.jpg,

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('../images/Object Detection/laptop.jpg')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Kurangi noise dengan GaussianBlur
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Deteksi tepi dengan Canny
edges = cv2.Canny(blur, 50, 150)

# Deteksi kontur
contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

output = img.copy()
cv2.drawContours(output, contours, -1, (0, 0, 255), 6) 

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title('Contour Detection (Canny + findContours)')
plt.axis('off')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import os

# Path folder berisi gambar wajah
folder_path = '../images/facedet'

# Load classifier Haar Cascade
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')


def deteksi_wajah(filename, scaleFactor=1.1, minNeighbors=8):
    img_path = os.path.join(folder_path, filename)

    if not os.path.exists(img_path):
        print(f"⚠️ File {filename} tidak ditemukan.")
        return

    # Baca dan ubah ke grayscale
    img = cv2.imread(img_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Deteksi wajah
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=scaleFactor,
        minNeighbors=minNeighbors,
        minSize=(30, 30)
    )

    # Gambar persegi di sekitar wajah
    img_detected = img.copy()
    for (x, y, w, h) in faces:
        cv2.rectangle(img_detected, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # Tampilkan hasil berdampingan (original & hasil deteksi)
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(img_detected, cv2.COLOR_BGR2RGB))
    plt.title(f'{filename} ({len(faces)} face(s))')
    plt.axis('off')

    plt.show()


# ===== Contoh pemanggilan =====
deteksi_wajah('prabowo.jpg', scaleFactor=1.1, minNeighbors=8)
deteksi_wajah('kartini.jpg', scaleFactor=1.1, minNeighbors=8)
deteksi_wajah('kucing.jpg', scaleFactor=1.1, minNeighbors=8)
deteksi_wajah('mask.png', scaleFactor=1.1, minNeighbors=8)
deteksi_wajah('mjordan.jpg', scaleFactor=1.15, minNeighbors=10)
deteksi_wajah('solvayconf.jpg', scaleFactor=1.05, minNeighbors=5)
```

## 4. Lakukan deteksi senyuman pada gambar people:

```python
import cv2
import matplotlib.pyplot as plt

# --- 1. Baca gambar ---
img_path = '../images/facedet/tawa.jpg'
img = cv2.imread(img_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# --- 2. Load Haar Cascade ---
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_smile.xml')

# --- 3. Deteksi wajah ---
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(75, 75)
)

# --- 4. Deteksi senyuman di tiap wajah ---
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 3)  # Kotak wajah
    
    roi_gray = gray[y:y+h, x:x+w]
    roi_color = img[y:y+h, x:x+w]

    smiles = smile_cascade.detectMultiScale(
        roi_gray,
        scaleFactor=1.04,
        minNeighbors=5,
        minSize=(25, 25)
    )

    for (sx, sy, sw, sh) in smiles:
        cv2.rectangle(roi_color, (sx, sy), (sx + sw, sy + sh), (255, 0, 0), 2)  # Kotak senyum

# --- 5. Tampilkan hasil ---
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(f'Detected Faces: {len(faces)}')
plt.axis('off')
plt.show()
```

## 5. Lakukan Face Tracking menggunakan Google Colab. Petunjuk, Tutorial selengkapnya tentang akses kamera dan FaceDetection

```python
import cv2

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray, 
        scaleFactor=1.1, 
        minNeighbors=5, 
        minSize=(30, 30)
    )

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)

    cv2.imshow('Face Tracking (Tekan q untuk keluar)', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

## 6. Lakukan Blurring pada bagian wajah yang terdeteksi.

```python
import cv2
import matplotlib.pyplot as plt

img_path = '../images/facedet/jokowi.jpg'
img = cv2.imread(img_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.5,
    minNeighbors=8,
    minSize=(75, 75)
)


for (x, y, w, h) in faces:
    face_roi = img[y:y+h, x:x+w]
    face_blur = cv2.medianBlur(face_roi, 45)
    img[y:y+h, x:x+w] = face_blur

plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(f'Blurred Faces ({len(faces)} face(s))')
plt.axis('off')
plt.show()

```

## 7. Mengolah gambar KTM masing-masing

```python
import cv2
import matplotlib.pyplot as plt

img_path = '../images/facedet/ktm.jpg'
img = cv2.imread(img_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.5,
    minNeighbors=7,
    minSize=(30, 30)
)

if len(faces) > 0:
    plt.figure(figsize=(10, 5))

    plt.subplot(1, len(faces) + 1, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('Gambar Asli')

    for i, (x, y, w, h) in enumerate(faces, start=2):
        face_crop = img[y:y+h, x:x+w]
        plt.subplot(1, len(faces) + 1, i)
        plt.imshow(cv2.cvtColor(face_crop, cv2.COLOR_BGR2RGB))
        plt.title(f'Face {i-1}')

    plt.tight_layout()
    plt.show()
else:
    print("Tidak ada wajah terdeteksi.")
```

### c. Lakukan Deteksi Obyek KTM pada gambar yang berisi KTM dengan barang lainnya dengan menggunakan Konsep Feature Matching

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

template_path = '../images/facedet/ktm.jpg'
scene_path = '../images/facedet/ktm_scene.jpg'

img_template = cv2.imread(template_path, cv2.IMREAD_GRAYSCALE)
img_scene = cv2.imread(scene_path, cv2.IMREAD_GRAYSCALE)
img_scene_color = cv2.imread(scene_path)

# 2. Inisialisasi ORB
orb = cv2.ORB_create(nfeatures=1000)

# 3. Deteksi keypoints dan deskriptor
kp1, des1 = orb.detectAndCompute(img_template, None)
kp2, des2 = orb.detectAndCompute(img_scene, None)

# 4. Matching menggunakan Brute-Force dengan Hamming distance
bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matches = bf.match(des1, des2)
matches = sorted(matches, key=lambda x: x.distance)

# 5. Ambil subset terbaik (misal 30 terbaik)
good_matches = matches[:30]

# 6. Gambar hasil matching
match_img = cv2.drawMatches(img_template, kp1, img_scene_color, kp2, good_matches, None, flags=2)

# 7. Estimasi posisi obyek dengan homography (jika cukup fitur cocok)
if len(good_matches) > 10:
    src_pts = np.float32([kp1[m.queryIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    dst_pts = np.float32([kp2[m.trainIdx].pt for m in good_matches]).reshape(-1, 1, 2)

    M, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)
    h, w = img_template.shape
    pts = np.float32([[0,0], [0,h], [w,h], [w,0]]).reshape(-1,1,2)
    dst = cv2.perspectiveTransform(pts, M)

    # Gambar poligon deteksi di scene
    img_scene_color = cv2.polylines(img_scene_color, [np.int32(dst)], True, (0,255,0), 3, cv2.LINE_AA)
else:
    print("Fitur tidak cukup untuk menemukan homografi")

# 8. Tampilkan hasil
plt.figure(figsize=(15, 6))
plt.subplot(1, 2, 1)
plt.imshow(match_img[..., ::-1])
plt.title('Feature Matching (ORB)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_scene_color, cv2.COLOR_BGR2RGB))
plt.title('Detected KTM in Scene')
plt.axis('off')
plt.tight_layout()
plt.show()
```
