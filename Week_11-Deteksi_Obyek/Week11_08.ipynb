```python
img_gray = cv.cvtColor(cats_and_bunnies, cv.COLOR_BGR2GRAY)
template_gray = cv.cvtColor(cat_template, cv.COLOR_BGR2GRAY)

w, h = template_gray.shape[::-1]

methods = [
    'cv.TM_SQDIFF',
    'cv.TM_SQDIFF_NORMED',
    'cv.TM_CCORR',
    'cv.TM_CCORR_NORMED',
    'cv.TM_CCOEFF',
    'cv.TM_CCOEFF_NORMED'
]

plt.figure(figsize=(10, 15))

for i, meth in enumerate(methods, 1):
    img_copy = cats_and_bunnies.copy()
    method = eval(meth)

    res = cv.matchTemplate(img_gray, template_gray, method)
    min_val, max_val, min_loc, max_loc = cv.minMaxLoc(res)

    if method in [cv.TM_SQDIFF, cv.TM_SQDIFF_NORMED]:
        top_left = min_loc
    else:
        top_left = max_loc
    bottom_right = (top_left[0] + w, top_left[1] + h)

    cv.rectangle(img_copy, top_left, bottom_right, (0, 0, 255), 2)

    img_show = cv.cvtColor(img_copy, cv.COLOR_BGR2RGB)

    plt.subplot(6, 2, (i - 1) * 2 + 1)
    plt.imshow(res, cmap='gray')
    plt.title(f'{meth} - Matching Result')
    plt.axis('off')

    plt.subplot(6, 2, (i - 1) * 2 + 2)
    plt.imshow(img_show)
    plt.title(f'{meth} - Detected point')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## 3. Implementasikan konsep template matching tanpa menggunakan library OpenCV untuk multiple object

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# --- Load & grayscale ---
img_path = '/home/dedybayu/Documents/PCVK_2025/images/Object Detection/bahrain.jpg'
tpl_path = '/home/dedybayu/Documents/PCVK_2025/images/Object Detection/bahrain-template.jpg'

img = np.array(Image.open(img_path).convert('L'), dtype=np.float32)
template = np.array(Image.open(tpl_path).convert('L'), dtype=np.float32)

img_h, img_w = img.shape
tpl_h, tpl_w = template.shape

# --- Normalisasi template ---
tpl_mean = template.mean()
tpl_std = template.std()
if tpl_std == 0:
    raise ValueError("Template has zero standard deviation")
tpl_norm = (template - tpl_mean) / tpl_std

# --- Hitung NCC (brute-force) ---
res_h = img_h - tpl_h + 1
res_w = img_w - tpl_w + 1
result = np.zeros((res_h, res_w), dtype=np.float32)

for y in range(res_h):
    for x in range(res_w):
        region = img[y:y+tpl_h, x:x+tpl_w]
        r_mean = region.mean()
        r_std = region.std()
        if r_std == 0:
            result[y, x] = 0.0
        else:
            region_norm = (region - r_mean) / r_std
            # dot product -> NCC numerator; denom already handled by normalization
            result[y, x] = (region_norm * tpl_norm).sum() / (tpl_h * tpl_w)  # /N to keep in [-1,1]

# normalisasi ke [0,1] (opsional tapi memudahkan threshold)
res_min, res_max = result.min(), result.max()
if res_max - res_min > 0:
    result_norm = (result - res_min) / (res_max - res_min)
else:
    result_norm = result.copy()

# --- Iterative detection + suppression ---
threshold = 0.65   # threshold
detections = []    # list of (x, y, score)

res_copy = result_norm.copy()

while True:
    idx_flat = np.argmax(res_copy)
    y_peak, x_peak = np.unravel_index(idx_flat, res_copy.shape)
    score = res_copy[y_peak, x_peak]
    if score < threshold:
        break
    detections.append((x_peak, y_peak, float(score)))

    y0 = max(0, y_peak - tpl_h // 2)
    y1 = min(res_copy.shape[0], y_peak + tpl_h // 2)
    x0 = max(0, x_peak - tpl_w // 2)
    x1 = min(res_copy.shape[1], x_peak + tpl_w // 2)
    res_copy[y0:y1, x0:x1] = 0.0



fig, ax = plt.subplots(2, 2, figsize=(15, 11))

# Input Image
ax[0][0].imshow(img, cmap='gray')
ax[0][0].set_title('Input image')

# Template Image
ax[0][1].imshow(template, cmap='gray')
ax[0][1].set_title('Template image')

# Matching map
ax[1][0].imshow(result_norm, cmap='hot')
ax[1][0].set_title('Matching map (normalized)')

ax[1][1].imshow(img, cmap='gray')
ax[1][1].set_title('Detected bounding boxes')
for (x, y, s) in detections:
    rect = plt.Rectangle((x, y), tpl_w, tpl_h, edgecolor='b', facecolor='none', linewidth=2)
    ax[1][1].add_patch(rect)
    ax[1][1].text(x, y-6, f'{s:.2f}', color='yellow', fontsize=9, backgroundcolor='black')

plt.tight_layout()
plt.show()

print("Detections (top-left x,y,score):")
for d in detections:
    print(d)
```

## 4. Implementasikan metode Sobel Edge Detection, Canny Edge Detection, dan Laplacian Edge Detection pada OpenCV

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar
img_car_park = cv.imread('/home/dedybayu/Documents/PCVK_2025/images/Object Detection/car-park.jpg')

# Konversi ke grayscale
gray = cv.cvtColor(img_car_park, cv.COLOR_BGR2GRAY)
```

### a. Sobel Edge Detection

```python
sobelx = cv.Sobel(gray, cv.CV_64F, 1, 0, ksize=3)
sobely = cv.Sobel(gray, cv.CV_64F, 0, 1, ksize=3)
sobel_combined = cv.magnitude(sobelx, sobely)
sobel_combined = cv.convertScaleAbs(sobel_combined)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1) 
plt.imshow(cv.cvtColor(img_car_park, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### b. Canny Edge Detection

```python
canny_edges = cv.Canny(gray, 100, 200)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1) 
plt.imshow(cv.cvtColor(img_car_park, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(canny_edges, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### c. Laplacian Edge Detection

```python
laplacian = cv.Laplacian(gray, cv.CV_64F)
laplacian = cv.convertScaleAbs(laplacian)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1) 
plt.imshow(cv.cvtColor(img_car_park, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 5. Implementasikan metode Harris Corner Detection dan Shi-Tomasi Detection pada OpenCV

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_chess_board = cv.imread('/home/dedybayu/Documents/PCVK_2025/images/Object Detection/chess-board.jpg')
gray = cv.cvtColor(img_chess_board, cv.COLOR_BGR2GRAY)
```

### a. Harris Corner Detection

```python
gray_float = np.float32(gray)
harris = cv.cornerHarris(gray_float, blockSize=2, ksize=3, k=0.04)
harris = cv.dilate(harris, None)  
img_chess_board_harris = img_chess_board.copy()
img_chess_board_harris[harris > 0.01 * harris.max()] = [0, 255, 0]

plt.figure(figsize=(10, 15))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(img_chess_board, cv.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(img_chess_board_harris, cv.COLOR_BGR2RGB))
plt.title('Harris Corner Detection (hijau)')

plt.tight_layout()
plt.show()
```

### b. Shi-Tomasi Detection

```python
corners = cv.goodFeaturesToTrack(gray, maxCorners=100, qualityLevel=0.01, minDistance=10)
corners = np.int32(corners)
img_chess_board_shi = img_chess_board.copy()
for c in corners:
    x, y = c.ravel()
    cv.circle(img_chess_board_shi, (x, y), 4, (0, 0, 255), -1)


plt.figure(figsize=(10, 15))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(img_chess_board, cv.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(img_chess_board_shi, cv.COLOR_BGR2RGB))
plt.title('Shi-Tomasi Corner Detection (merah)')

plt.tight_layout()
plt.show()
```

## 6. Implementasikan metode Hough Transform pada OpenCV dengan menggunakan gambar sudoku.jpg.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar
img = cv.imread('/home/dedybayu/Documents/PCVK_2025/images/Object Detection/sudoku.jpg')
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# --- 1. Deteksi tepi menggunakan Canny ---
edges = cv.Canny(gray, 50, 150, apertureSize=3)

# --- 2. Hough Transform (Standard) ---
lines = cv.HoughLines(edges, 1, np.pi / 180, 150)

img_hough = img.copy()
if lines is not None:
    for line in lines:
        rho, theta = line[0]
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * rho
        y0 = b * rho
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))
        cv.line(img_hough, (x1, y1), (x2, y2), (0, 0, 255), 2)

# --- 3. Hough Transform Probabilistik (garis putus-putus lebih pendek) ---
linesP = cv.HoughLinesP(edges, 1, np.pi / 180, threshold=100, minLineLength=80, maxLineGap=10)
img_houghP = img.copy()
if linesP is not None:
    for line in linesP:
        x1, y1, x2, y2 = line[0]
        cv.line(img_houghP, (x1, y1), (x2, y2), (0, 255, 0), 2)

# --- 4. Tampilkan hasil ---
plt.figure(figsize=(15, 8))

plt.subplot(1, 3, 1)
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv.cvtColor(img_houghP, cv.COLOR_BGR2RGB))
plt.title('Hough Transform (Probabilistic)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 7. Implementasikan fungsi findContours() pada OpenCV untuk contour detection dengan menggunakan gambar laptop.jpg,

```python
import cv2

# 1. Baca gambar
img = cv2.imread('/home/dedybayu/Documents/PCVK_2025/images/Object Detection/laptop.jpg')

# 2. Ubah ke grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. Gunakan threshold untuk memisahkan objek dan background
_, thresh = cv2.threshold(gray, 100, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# 4. Temukan kontur menggunakan findContours()
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# 5. Gambar semua kontur pada gambar asli
output = img.copy()
cv2.drawContours(output, contours, -1, (0, 255, 0), 2)  # warna hijau, tebal 2px

# 6. Tampilkan hasil
plt.imshow(output)
plt.show()
```
